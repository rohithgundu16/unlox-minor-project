### Q3. Why does `0.1 + 0.2 == 0.3` return `False` in Python?

It returns `False` because Python stores decimal numbers in binary form, and some decimal values cannot be stored exactly. So the result of `0.1 + 0.2` is very close to `0.3`, but not exactly equal to it.

This is why direct comparison of some floating-point values can give `False`  result.

```python
print(0.1 + 0.2 == 0.3)  # False
```

A list is better when we need to add, remove, or update items. A tuple is useful when the data should stay the same.

### Q5. Why is dictionary lookup usually faster than searching in a list?

dictionary stores values using keys,(key-value pairs) so Python can usually find a value directly using its key. This takes about **O(1)** time on average.

In a list, Python may need to check each item one by one until it finds the required value. This takes **O(n)** time in the worst case. means we can check every element present in the list. So dictionary is usually faster.

### Q7. What is the difference between `append()`, `extend()`, and `insert()`?

All three are used to add items to a list, but they work differently:

- `append()` adds the given item at the end of the list. If we give it another list, that whole list is added as one item.
- `extend()` adds each item from another iterable to the end of the list.
- `insert()` adds an item at the position we specify.

```python
numbers = [1, 2, 3]

numbers.append([4, 5])
print(numbers)  # [1, 2, 3, [4, 5]]

numbers = [1, 2, 3]
numbers.extend([4, 5])
print(numbers)  # [1, 2, 3, 4, 5]

numbers = [1, 2, 3]
numbers.insert(1, 10)
print(numbers)  # [1, 10, 2, 3]
```

### Q8. Explain unpacking with `*` in Python. What happens if there are not enough values?

The `*` operator collects the remaining values into a list while unpacking.

For example, in `a, *b, c = [10, 20, 30, 40, 50]`, the first value goes to `a`, the last value goes to `c`, and the middle values go into `b`.

If there are not enough values to assign to the normal variables, Python gives a `ValueError`.

```python
a, *b, c = [10, 20, 30, 40, 50]
print(a)  # 10
print(b)  # [20, 30, 40]
print(c)  # 50

# This gives ValueError because there are not enough values:
# a, *b, c = [10]
```

### Q9. Explain `/`, `//`, `%`, and `**` using `-7` and `2`.

These operators are used for different mathematical operations:

- `/` It gives normal division.
- `//` It gives floor division, Means it rounds down to the next lower integer.
- `%` gives the remainder.
- `**` It is used to gives the  power of any number.

```python
a=7
print(a/2) # 3.5
print(a//2) # 3
print(a%2) # 1
print(a**2) # 49
```
