Q17. Swiggy Order Calculator.

Q18. Temperature Converter.
Ask user for a value and unit (C or F). Convert and print result rounded to 2 decimal places. Handle invalid unit input gracefully (print an error message, do NOT crash).

```python
temperature = float(input("Enter temperature: "))
unit = input("Enter unit (C/F): ").upper()

if unit == "C":
    result = (temperature * 9 / 5) + 32
    print("Temperature in Fahrenheit:", round(result, 2))
elif unit == "F":
    result = (temperature - 32) * 5 / 9
    print("Temperature in Celsius:", round(result, 2))
else:
    print("Invalid unit. Please enter C or F.")
```

Q24. Cricket Strike Rate Classifier.
Take runs scored and balls faced. Compute strike rate = (runs/balls) × 100. Classify:
- SR > 150 → Explosive (T20 finisher)
- 100 < SR ≤ 150 → Aggressive (Top order T20)
- 75 ≤ SR ≤ 100 → Steady (ODI anchor)
- SR < 75 → Defensive (Test match)
Edge case: handle balls = 0 (do not crash, print message instead).

```python
runs = int(input("Enter runs scored: "))
balls = int(input("Enter balls faced: "))

if balls == 0:
    print("Cannot calculate strike rate when balls faced is zero")
else:
    strike_rate = (runs / balls) * 100
    print("Strike rate:", round(strike_rate, 2))

    if strike_rate > 150:
        print("Explosive (T20 finisher)")
    elif strike_rate > 100:
        print("Aggressive (Top order T20)")
    elif strike_rate >= 75:
        print("Steady (ODI anchor)")
    else:
        print("Defensive (Test match)")
```

Print this number diamond for N = 5 (the digits go up to N in the middle row):

```python
n = 5

for i in range(1, n + 1):
    for space in range(n - i):
        print(" ", end="")
    for j in range(1, i + 1):
        print(j, end=" ")
    for j in range(i - 1, 0, -1):
        print(j, end=" ")
    print()

for i in range(n - 1, 0, -1):
    for space in range(n - i):
        print(" ", end="")
    for j in range(1, i + 1):
        print(j, end=" ")
    for j in range(i - 1, 0, -1):
        print(j, end=" ")
    print()
```

Q31.
Find all Armstrong numbers between 1 and 9999. (An n-digit number is Armstrong if the sum of its digits each raised to the power n equals the number itself. Example: 153 = 1³ + 5³ + 3³.) Print them in a clean comma-separated list with a count at the end.

```python
armstrong_numbers = []

for number in range(1, 10000):
    temp = number
    digit_count = 0
    while temp > 0:
        digit_count += 1
        temp //= 10

    temp = number
    total = 0
    while temp > 0:
        digit = temp % 10
        total += digit ** digit_count
        temp //= 10

    if total == number:
        armstrong_numbers.append(number)

print("Armstrong numbers:", ", ".join(map(str, armstrong_numbers)))
print("Count:", len(armstrong_numbers))
```

Q33. Prime Check, optimized
Take a number N. Check if it is prime by iterating only up to √N. Then, print the SUM of all primes between 2 and N (inclusive). Do not use any library.

```python
n = int(input("Enter N: "))

if n < 2:
    print("No primes in this range")
    print("Sum of primes: 0")
else:
    prime_sum = 0
    for number in range(2, n + 1):
        is_prime = True
        divisor = 2

        while divisor * divisor <= number:
            if number % divisor == 0:
                is_prime = False
                break
            divisor += 1

        if is_prime:
            prime_sum += number

    print("Sum of primes:", prime_sum)
```

Q34.
Take a number as input. Reverse it WITHOUT converting to string. Then check if the original and reversed are equal (palindrome). Handle negative numbers, they should be considered NOT palindrome (because of the minus sign). Examples: 121 → palindrome, 123 → not palindrome, -121 → not palindrome.

```python
number = int(input("Enter a number: "))

if number < 0:
    print("Reversed number:", number)
    print("Not a palindrome")
else:
    original = number
    temp = number
    reverse = 0

    while temp > 0:
        digit = temp % 10
        reverse = reverse * 10 + digit
        temp //= 10

    print("Reversed number:", reverse)
    if original == reverse:
        print("Palindrome")
    else:
        print("Not a palindrome")
```

Q36. String Scanner (Single Pass).
Take a sentence (e.g., a Swiggy review like "Order came on time, food was hot, packaging 10/10!"). In ONE pass through the string, count: total characters, total words, vowels (a, e, i, o, u - case-insensitive), consonants, digits, spaces, and special characters. Print all counts.

```python
sentence = input("Enter a sentence: ")

total_characters = len(sentence)
words = vowels = consonants = digits = spaces = special = 0
inside_word = False

for ch in sentence:
    if ch.isalpha():
        if ch.lower() in "aeiou":
            vowels += 1
        else:
            consonants += 1
        if not inside_word:
            words += 1
        inside_word = True
    elif ch.isdigit():
        digits += 1
        if not inside_word:
            words += 1
        inside_word = True
    elif ch.isspace():
        spaces += 1
        inside_word = False
    else:
        special += 1
        inside_word = False

print("Total characters:", total_characters)
print("Total words:", words)
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", special)
```

Q37.
Take 10 student marks (out of 100) as input, space-separated on one line, parse with split() and map(). Without using sort() or sorted(), find: (a) maximum, (b) minimum, (c) average, (d) count of students above average, (e) the SECOND highest mark.


```python
marks = list(map(int, input("Enter 10 marks separated by spaces: ").split()))

if len(marks) != 10:
    print("Please enter exactly 10 marks")
else:
    maximum = marks[0]
    minimum = marks[0]
    total = 0

    for mark in marks:
        if mark > maximum:
            maximum = mark
        if mark < minimum:
            minimum = mark
        total += mark

    average = total / len(marks)
    above_average = 0
    second_highest = None

    for mark in marks:
        if mark > average:
            above_average += 1
        if mark < maximum:
            if second_highest is None or mark > second_highest:
                second_highest = mark

    print("Maximum:", maximum)
    print("Minimum:", minimum)
    print("Average:", round(average, 2))
    print("Students above average:", above_average)
    print("Second highest distinct mark:", second_highest)
```
